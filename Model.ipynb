##### Performance metric


```python
from ultralytics import YOLO

# Pretrained model load karo
model = YOLO("/home/vedant/MINOR PROJECT/PPE KIT/runs/detect/train-4/weights/best.pt")

# Apne dataset par evaluation run karo
metrics = model.val(data="/home/vedant/MINOR PROJECT/PPE KIT/20250731-ppe2286y/data.yaml")

# Metrics access karne ke liye:
print("mAP50-95:", metrics.box.map)
print("mAP50:", metrics.box.map50)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt

# 1. Model load karein (path mein 'train-' ki jagah sahi folder name daalein)
model = YOLO("/home/vedant/MINOR PROJECT/PPE KIT/runs/detect/train-14/weights/best.pt")

# 2. Pure folder par predict karein aur save=True set karein
results = model.predict(
    source="/home/vedant/MINOR PROJECT/test",
    iou = 0.4,
    save=True  # Saari output images 'runs/detect/predict' folder mein save ho jaayengi
)

# 3. Notebook mein ek-ek karke saari images display karne ke liye loop:
for i, r in enumerate(results):
    plt.figure(figsize=(6, 6))
    plt.imshow(r.plot()[:, :, ::-1])
    plt.title(f"Result {i+1}")
    plt.axis("off")
    plt.show()
```
